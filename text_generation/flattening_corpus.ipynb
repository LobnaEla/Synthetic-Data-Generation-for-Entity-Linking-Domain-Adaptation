```python
import json
```

```python
def flatten_corpus(original_corpus):
    """
    Découpe le corpus pour obtenir un objet unique par mention.
    Si un paragraphe contient 3 mentions, il sera dupliqué 3 fois,
    chacune associée à sa mention respective et à ses 100 candidats.
    """
    flat_corpus = []
    
    for item in original_corpus:
        paragraph_text = item.get("paragraph", "")
        mentions = item.get("mentions", [])
        
        for mention in mentions:
            # On construit un nouvel objet isolé pour CETTE mention spécifique
            single_mention_data = {
                "paragraph": paragraph_text,
                # On extrait les infos de la mention unique
                "mention": mention.get("mention"),
                "start": mention.get("start"),
                "end": mention.get("end"),
                "entity_id": mention.get("entity_id"), # La bonne réponse 
            }
            flat_corpus.append(single_mention_data)
            
    return flat_corpus
```

```python
def load_corpus_from_jsonl(file_path):
    """
    Charge le corpus depuis un fichier JSONL.
    Chaque ligne du fichier doit être un objet JSON représentant un paragraphe avec ses mentions.
    """
    import json
    
    corpus = []
    try:
        with open(file_path, 'r', encoding='utf-8') as f:
            for line in f:
                item = json.loads(line.strip())
                corpus.append(item)
    except Exception as e:
        print(f"❌ Une erreur est survenue lors de la lecture : {e}")
    
    return corpus
```

```python
# 1. Votre corpus généré par le LLM et enrichi en candidats (ex: 5 paragraphes de test)
corpus = load_corpus_from_jsonl("output/corpus_Historic_blogs.jsonl")  # Remplacez par le domaine que vous souhaitez traiter

# 2. Application du découpage
corpus_flat = flatten_corpus(corpus)

print(f"📊 Statistiques du découpage :")
print(f"   • Paragraphes d'origine : {len(corpus)}")
print(f"   • Échantillons finaux (1 paragraphe / 1 mention) : {len(corpus_flat)}")

# 3. Sauvegarde au format JSONL (une ligne par mention cible)
output_flat_path = "corpus_Historic_blogs_flat.jsonl"

with open(output_flat_path, "w", encoding="utf-8") as f:
    for item in corpus_flat:
        f.write(json.dumps(item, ensure_ascii=False) + "\n")

print(f"\n💾 Fichier enregistré avec succès dans : {output_flat_path}")
```
